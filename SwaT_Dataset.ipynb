```python
import matplotlib.pyplot as plt

# Example data
time = [1, 2, 3, 4, 5]
flow_rate = [10, 20, 15, 25, 30]

plt.figure(figsize=(6,4))
plt.plot(time, flow_rate, marker='o', color='b', label='Flow Rate')
plt.title("Line Chart - Flow Rate Over Time")
plt.xlabel("Time")
plt.ylabel("Flow Rate")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Example data
valves = ['MV101', 'MV201', 'MV301', 'MV401']
status = [5, 7, 3, 6]

plt.figure(figsize=(6,4))
plt.bar(valves, status, color='teal')
plt.title("Bar Chart - Valve Activation Count")
plt.xlabel("Valve ID")
plt.ylabel("Count")
plt.show()
```

```python
# Example data
labels = ['Normal', 'Attack']
sizes = [80, 20]

plt.figure(figsize=(5,5))
plt.pie(sizes, labels=labels, autopct='%1.1f%%', startangle=140, colors=['lightgreen', 'salmon'])
plt.title("Pie Chart - Normal vs Attack Data")
plt.show()
```
